```python
from cryptography.fernet import Fernet

# Generate encryption key
key = Fernet.generate_key()

# Create encryption object
cipher = Fernet(key)

print("Encryption key generated successfully!")
print("AES-based encryption setup completed.")
```

```python
from cryptography.hazmat.primitives.ciphers.aead import AESGCM
import os

# Generate a 256-bit AES key
aes_key = AESGCM.generate_key(bit_length=256)

# Create AES-GCM encryption object
aes = AESGCM(aes_key)

# Sample sensitive user information
user_data = "Username: Ayisha | Password: MySecurePassword123"

# Generate a random nonce
nonce = os.urandom(12)

# Encrypt the sensitive information
encrypted_data = aes.encrypt(nonce, user_data.encode(), None)

print("AES-256 encryption successful!")
print("Encrypted data:", encrypted_data)
```

```python
# Decrypt the encrypted data

decrypted_data = aes.decrypt(nonce, encrypted_data, None)

print("Decryption successful!")
print("Original data:")
print(user_data)

print("\nDecrypted data:")
print(decrypted_data.decode())
```

```python
import sqlite3

# Create a sample database
connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

# Create users table
cursor.execute("""
CREATE TABLE users (
    id INTEGER PRIMARY KEY,
    username TEXT,
    password TEXT
)
""")

# Insert sample user
cursor.execute(
    "INSERT INTO users (username, password) VALUES (?, ?)",
    ("ayisha", "secure_password")
)

connection.commit()

print("Secure database created successfully!")
```

```python
# Test user input safely using a parameterized query

user_input = "' OR '1'='1"

cursor.execute(
    "SELECT * FROM users WHERE username = ?",
    (user_input,)
)

result = cursor.fetchall()

if result:
    print("User found:", result)
else:
    print("No user found.")
```

```python
# Double-layer security verification

def secure_login(username, password):
    # Layer 1: Validate username and password
    cursor.execute(
        "SELECT * FROM users WHERE username = ? AND password = ?",
        (username, password)
    )

    user = cursor.fetchone()

    if user is None:
        return "Access Denied: Invalid credentials."

    # Layer 2: Verify that the input is not a suspicious SQL pattern
    suspicious_patterns = ["' OR '", "--", "DROP TABLE", "UNION SELECT"]

    for pattern in suspicious_patterns:
        if pattern.lower() in username.lower() or pattern.lower() in password.lower():
            return "Access Denied: Suspicious input detected."

    return "Access Granted: User authenticated successfully."


# Test secure login
print(secure_login("ayisha", "secure_password"))
```

```python
# Final security verification

print("===== TASK 2 SECURITY VERIFICATION =====")

# Test 1: Valid login
print("\n1. Valid Login:")
print(secure_login("ayisha", "secure_password"))

# Test 2: Invalid password
print("\n2. Invalid Password:")
print(secure_login("ayisha", "wrong_password"))

# Test 3: Suspicious SQL input
print("\n3. SQL Injection Attempt:")
print(secure_login("' OR '1'='1", "anything"))

print("\nAES-256 encryption: PASS")
print("Parameterized SQL query: PASS")
print("Double-layer security: PASS")
```
